## Q6. Sum of numbers divisible by both 7 and 9

```python
n = int(input("Enter n: "))
total = 0
for number in range(1, n + 1):
    if number % 7 == 0 and number % 9 == 0:
        total += number
print("Sum:", total)
```

## Q7. Sum of prime numbers from 1 to n

```python
n = int(input("Enter n: "))
prime_sum = 0
for number in range(2, n + 1):
    is_prime = True
    for divisor in range(2, int(number ** 0.5) + 1):
        if number % divisor == 0:
            is_prime = False
            break
    if is_prime:
        prime_sum += number
print("Sum of primes:", prime_sum)
```

## Q8. Sum of odd numbers from 1 to n using a function

```python
def sum_odd(n):
    total = 0
    for number in range(1, n + 1, 2):
        total += number
    return total

n = int(input("Enter n: "))
print("Sum of odd numbers:", sum_odd(n))
```

## Q9. Sum of prime numbers from 1 to n using a function

```python
def sum_primes(n):
    total = 0
    for number in range(2, n + 1):
        prime = True
        for divisor in range(2, int(number ** 0.5) + 1):
            if number % divisor == 0:
                prime = False
                break
        if prime:
            total += number
    return total

n = int(input("Enter n: "))
print("Sum of prime numbers:", sum_primes(n))
```
